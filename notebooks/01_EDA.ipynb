```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

#Display Settings

pd.set_option('display.max_columns', None)
sns.set_style('whitegrid')
```

```python
#Load Training Data

df = pd.read_csv("../data/raw/train.csv")
print(f'Training Data Shape: {df.shape}')
df.head()
```

```python
df.describe()
```

```python
#Missign Value Analysis

missing = df.isnull().sum()
missing_pct = (missing / len(df)) * 100
missing_df = pd.DataFrame({'Missing Values': missing, 'Percentage': missing_pct})
missing_df = missing_df[missing_df['Missing Values'] > 0].sort_values("Percentage", ascending=False)
print(f'Total Columns with Missing Values: {missing_df.shape[0]}')
missing_df
```

```python
#Target Variable Distribution
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

#Original Distribution
axes[0].hist(df['SalePrice'], bins=50, color='steelblue', edgecolor='black')
axes[0].set_title('Original SalePrice Distribution')
axes[0].set_xlabel('Price ($)')
axes[0].set_ylabel('Count')

#Log Transformed Distribution
axes[1].hist(np.log1p(df['SalePrice']), bins=50, color='coral', edgecolor='black')
axes[1].set_title('Log Transformed SalePrice Distribution')
axes[1].set_xlabel('Log(Price)')
axes[1].set_ylabel('Count')

plt.tight_layout()
plt.savefig('../reports/figures/saleprice_distribution.png', dpi=150)
plt.show()
```

```python
#TOP 15 features Correlated with SalePrice

correlation = df.corr(numeric_only=True)['SalePrice'].sort_values(ascending=False)
top15 = correlation.head(16)  # 16 because SalePrice itself is first
print("Top 15 Features Correlated with SalePrice:\n")
print(top15)
```

```python
#Correlation Heatmap for top 15 features

top_features = top15.index.tolist()
plt.figure(figsize=(12, 10))
sns.heatmap(
    df[top_features].corr(), 
    annot=True, #Show numbers on the map
    fmt=".2f", # 2 decimal places
    cmap='coolwarm', #Blue-Red color scheme
    square=True
)
plt.title('Correlation Heatmap of Top 15 Features', fontsize=14)
plt.tight_layout()
plt.savefig('../reports/figures/correlation_heatmap.png', dpi=150)
plt.show()
```

```python
#Scatter Plots: Top Features vs SalePrice (Outlier Detection)

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

#Overall Quality vs SalePrice
axes[0, 0].scatter(df['OverallQual'], df['SalePrice'], alpha=0.5, color='steelblue')
axes[0, 0].set_title('OverallQual vs SalePrice')
axes[0, 0].set_xlabel('Overall Quality')
axes[0, 0].set_ylabel('SalePrice ($)')

#GrLivArea vs SalePrice
axes[0, 1].scatter(df['GrLivArea'], df['SalePrice'], alpha=0.5, color='coral')
axes[0, 1].set_title('GrLivArea vs SalePrice')
axes[0, 1].set_xlabel('Living Area (sqft)')
axes[0, 1].set_ylabel('SalePrice ($)')

#YearBuilt vs SalePrice
axes[1, 0].scatter(df['YearBuilt'], df['SalePrice'], alpha=0.5, color='seagreen')
axes[1, 0].set_title('YearBuilt vs SalePrice')
axes[1, 0].set_xlabel('Year Built')
axes[1, 0].set_ylabel('SalePrice ($)')

#TotalBsmtSF vs SalePrice
axes[1, 1].scatter(df['TotalBsmtSF'], df['SalePrice'], alpha=0.5, color='mediumpurple')
axes[1, 1].set_title('TotalBsmtSF vs SalePrice')
axes[1, 1].set_xlabel('Total Basement (sqft)')
axes[1, 1].set_ylabel('SalePrice ($)')

plt.tight_layout()
plt.savefig('../reports/figures/scatter_plots_top_features.png', dpi=150)
plt.show()
```

```python
#Categorical features vs SalePrice

fig,axes = plt.subplots(2, 2, figsize=(16, 10))

#Neighborhood(top factor for location)
neighborhood_median = df.groupby('Neighborhood')['SalePrice'].median().sort_values()
axes[0, 0].barh(neighborhood_median.index, neighborhood_median.values, color='steelblue')
axes[0, 0].set_title('Median SalePrice by Neighborhood')
axes[0, 0].set_xlabel('Median SalePrice ($)')

#House Style
sns.boxplot(x='HouseStyle', y='SalePrice', data=df, ax=axes[0, 1], palette='Set2')
axes[0, 1].set_title('SalePrice by House Style')
axes[0, 1].tick_params(axis='x', rotation=45)

#Sale Condition
sns.boxplot(data=df, x='SaleCondition', y='SalePrice', ax=axes[1, 0], palette='Set3')
axes[1, 0].set_title('SalePrice by Sale Condition')

#Overall Quality (as categorical)
sns.boxplot(data=df, x='OverallQual', y='SalePrice', ax=axes[1, 1], palette='RdYlGn')
axes[1, 1].set_title('SalePrice by Overall Quality')

plt.tight_layout()
plt.savefig('../reports/figures/categorical_analysis.png', dpi=150)
plt.show()
```

## EDA Summary

### Key Findings:
1. **SalePrice** is right-skewed → will use log transform
2. **OverallQual** (0.79) and **GrLivArea** (0.71) are the strongest predictors
3. **Neighborhood** has huge impact on price (6x difference)
4. **GarageCars & GarageArea** are highly correlated (0.88) → keep only one
5. **Outliers detected:** 2 houses with GrLivArea > 4000 but low price → remove
